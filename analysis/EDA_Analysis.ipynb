```python
# Drop unnecessary columns, such as 'Order Zipcode' and 'Product Description', which may not be relevant for analysis
df.drop(columns=['Order Zipcode','Product Description'],inplace=True)
```

```python
# Drop rows with missing values
df.dropna(inplace=True)
```

```python
# Identify text columns
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

print(text_columns)
```

```python
# Remove leading and trailing whitespace from text columns
for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
# Drop unnecessary columns
df.drop(columns=[
    "Customer Email",
    "Customer Password",
    "Customer Fname",
    "Customer Lname",
    "Customer Street",
    "Customer Zipcode",
    "Product Image"
], inplace=True)
```

```python
# Convert date columns to datetime format
df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"],
    errors="coerce"
)

df["shipping date (DateOrders)"] = pd.to_datetime(
    df["shipping date (DateOrders)"],
    errors="coerce"
)
```

```python
#  Check outliers in the dataset using boxplots for numerical columns
numerical_columns = df.select_dtypes(include=[np.number]).columns
for col in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
# 1. Monthly Sales Trend
df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"],
    errors="coerce"
)
monthly_sales = (
    df.groupby(
        df["order date (DateOrders)"].dt.to_period("M")
    )["Sales"]
    .sum()
    .reset_index()
)
monthly_sales["Month"] = monthly_sales[
    "order date (DateOrders)"
].astype(str)
```

```python
# Plotting the monthly sales trend
plt.figure(figsize=(14, 6))
plt.plot(
    monthly_sales["Month"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=2
)
plt.title(
    "Monthly Sales Trend",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Sales by Category
category_sales = (
    df.groupby("Category Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
# Plotting sales by product category
plt.figure(figsize=(12, 7))
category_sales.plot(
    kind="bar"
)
plt.title(
    "Sales by Product Category",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Insight into Top Sales Categories
print("Top Sales Category:")
print(category_sales.head(5))
```

```python
# Sales by Market
market_sales = (
    df.groupby("Market")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
# Plotting sales by market
plt.figure(figsize=(11, 6))
market_sales.plot(
    kind="bar"
)
plt.title(
    "Sales by Market",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Market")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Insight into Market Performance
print("Market Performance:")
print(market_sales)
```

```python
# Sales by Customer Segment
segment_sales = (
    df.groupby("Customer Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
# Plotting sales by customer segment
plt.figure(figsize=(10, 6))
segment_sales.plot(
    kind="bar"
)
plt.title(
    "Sales by Customer Segment",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Insight into Customer Segment Performance
print("Customer Segment Performance:")
print(segment_sales)
```

```python
# Delivery Status
Delivery_Status = df["Delivery Status"].value_counts()
```

```python
# Plotting delivery status distribution
plt.figure(figsize=(8, 5))
plt.bar(Delivery_Status.index, Delivery_Status.values)
plt.xlabel("Delivery Status")
plt.ylabel("Count")
plt.title("Distribution of Delivery Status")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Insight into Delivery Status
print(Delivery_Status)
```

```python
# Late Delivery Risk by Shipping Mode
Late_Risk_by_Shipping_Mode = (df.groupby("Shipping Mode")["Late_delivery_risk"].mean().mul(100).sort_values(ascending=False))
```

```python
# Plotting delivery status distribution
plt.figure(figsize=(8, 5))
Late_Risk_by_Shipping_Mode.plot(kind='bar')
plt.title('Late Delivery Risk by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Late Delivery Risk (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Insight into late delivery risk by shipping mode
print(Late_Risk_by_Shipping_Mode)
```

```python
# Profit by Product Category
Profit_by_Category = (df.groupby("Category Name")["Benefit per order"].sum().sort_values(ascending=False))
```

```python
# Plotting Benefit per order by product category
plt.figure(figsize=(25,10))
Profit_by_Category.plot(kind='bar')
plt.title('Profit_By_Category',size=16,fontweight='bold')
plt.xlabel('Category Name',size=16,fontweight='bold')
plt.ylabel('Total profit',size=16,fontweight='bold')
plt.xticks(rotation=45)
plt.yticks(fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Insight into Profit by Product Category
# This is important because a category with high sales isn't necessarily the category with the highest profit.
print("Top 5 Categories by Profit:")
print(Profit_by_Category.head(5))
```

```python
# Sales vs Profit Relationship
plt.figure(figsize=(10, 7))

plt.scatter(
    df["Sales"],
    df["Order Profit Per Order"],
    alpha=0.3
)
plt.title(
    "Sales vs Profit per Order",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Sales")
plt.ylabel("Profit per Order")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

This helps identify: 

high-sales/high-profit orders
high-sales/low-profit orders
